```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
```

```python
#Метод Ейлера-Коші
def f(x, y):
    return x + np.cos(y / 2.25)

a=1.4 # ліва межа відрізку
b=2.4 # права межа відрізку
h=0.1 # крок
y0=2.2 # початкова умова
n=int((b-a)/h) # кількість кроків

# Генерація х, враховуючи останнє значення
x= np.arange(a, b +h, h) # Задати х з включенням в

y=np.empty(n+1)
y[0]=y0

# Метод Ейлера-Коші
for i in range(n):
    # Використовується середнє значення для нахилу
    y[i+1]=y[i]+(f(x[i],y[i])+f(x[i+1],y[i]+h*f(x[i],y[i])))*h/2

y_rounded = np.round(y, 4)
print(" X=",x, "\ny=", y_rounded)

# Візуалізація
plt.plot(x, y, "o", x, y, "red")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Метод Ейлера-Коші")
plt.legend(["точки", "x+cos(y/2.25)"])
plt.grid()
plt.show()
```

```python
#Метод Ейлера-Коші
def f(x, y):
    return x + np.sin(y / np.sqrt(10))

a=0.6 # ліва межа відрізку
b=1.6 # права межа відрізку
h=0.1 # крок
y0=0.8 # початкова умова
n=int((b-a)/h) # кількість кроків

# Генерація х, враховуючи останнє значення
x= np.arange(a, b +h, h) # Задати х з включенням в

y=np.empty(n+1)
y[0]=y0

# Метод Ейлера-Коші
for i in range(n):
    # Використовується середнє значення для нахилу
    y[i+1]=y[i]+(f(x[i],y[i])+f(x[i+1],y[i]+h*f(x[i],y[i])))*h/2

y_rounded = np.round(y, 4)
print(" X=",x, "\ny=", y_rounded)

# Візуалізація
plt.plot(x, y, "o", x, y, "red")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Метод Ейлера-Коші")
plt.legend(["точки", "x+sin(y/sqrt(10))"])
plt.grid()
plt.show()
```

```python
# Функція, що повертає dy/dx
def model(x, y):
    return x + np.sin(y / np.sqrt(10))

# Початкова умова
y0=[0.8] # Потрібно передати як список

# Значення х
x= np.linspace(0.6, 1.6, 11)

# Розв'язання ODE
sol = solve_ivp(model, [0.6, 1.6], y0, t_eval=x)

# Виведення результатів
print(' x=', sol.t)
print(' y=', sol.y[0])

# Побудова графіка результатів
plt.plot(sol.t, sol.y[0], marker ='o')
plt.xlabel('x')
plt.ylabel('y(x)')
plt.title('Розв\'язання диф. рівняння')
plt.grid()
plt.show()
```
